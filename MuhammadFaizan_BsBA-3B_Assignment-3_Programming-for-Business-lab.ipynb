```python
from datetime import datetime ## We are using datetime to record the transaction history.
USERNAME = "faizan"
PIN = "1234"
balance = 1000.0
history = []
def login(): ## this function is defined to check whether the user login id and password is correct or not
    attempts = 3 ## i have placed this attempt variable so we can limit the login atempt if there are more failed attempts then the account can be locked from illegal usage
    while attempts > 0:
        user = input("Username: ").strip().lower() ## this is placed so the user id can be entered correct case.
        pin = input("PIN: ")
        if user == USERNAME and pin == PIN:
            print("Login successful. Welcome, Muhammad Faizan!")
            return True ## we are using return true because it highlights that the provided input matches the correct format.
        attempts -= 1
        print("Wrong username or PIN. Attempts left:", attempts)
    print("Too many failed attempts. Account locked.")
    return False 
def check_balance(): ## this is created so the user can check how much the amount is remaining in the users account.
    print("Current balance:", balance)
def deposit(): ## this is created so user can add more amount to his/her account.
    global balance ## balance is created outside every function, so it’s a global variable. A function can read a global variable freely, but if it tries to change it, Python assumes you’re making a brand new local variable with the same name
    try:
        amount = float(input("Enter amount to deposit: "))
        if amount <= 0:
            print("Amount must be greater than zero.")
        else:
            balance += amount
            time_now = datetime.now().strftime("%d-%m-%Y %H:%M:%S") ## This asks your computer for the current date and time, The name stands for “string format time”. It converts the datetime object into text, in the layout you choose. The characters inside the brackets describe that layout.
            history.append(time_now + " | Deposit: +" + str(amount)) ## This stores the finished text in a variable, which we then join to the transaction.
            print("Deposited:", amount)
            print("New balance:", balance)
    except ValueError:
        print("Invalid input. Please enter a number.")
def withdraw(): 
    global balance
    try:
        amount = float(input("Enter amount to withdraw: "))
        if amount <= 0:
            print("Amount must be greater than zero.")
        elif amount > balance:
            print("Insufficient balance.")
        else:
            balance -= amount
            time_now = datetime.now().strftime("%d-%m-%Y %H:%M:%S")
            history.append(time_now + " | Withdraw: -" + str(amount))
            print("Withdrawn:", amount)
            print("New balance:", balance)
    except ValueError:
        print("Invalid input. Please enter a number.")
def show_history():
    if len(history) == 0:
        print("No transactions yet.")
    else:
        print("\t Transaction History ")
        for item in history:
            print(item)
def bank_system():
    if not login():
        return
    while True:
        print("\t BANK MENU ")
        print("1. Check Balance")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. Transaction History")
        print("5. Exit")
        choice = input("Choose an option (1-5): ")
        if choice == "1":
            check_balance()
        elif choice == "2":
            deposit()
        elif choice == "3":
            withdraw()
        elif choice == "4":
            show_history()
        elif choice == "5":
            print("Thank you for banking with us. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.") 
bank_system()
```

The final version of the code


```python
import json
from datetime import datetime
USERNAME = "faizan"
PIN = "1234"
balance = 1000.0
history = []
FILE_NAME = "bank_data.json"
def load_data(): ## in this code we have created a json file to store the activities of our code
    global balance, history
    try:
        file = open(FILE_NAME, "r")
        data = json.load(file)
        file.close()
        balance = data["balance"]
        history = data["history"]
        print("Previous records loaded.")
    except FileNotFoundError:
        print("No saved data found. Starting fresh.")
    except (json.JSONDecodeError, KeyError):
        print("Data file is damaged. Starting fresh.")
def save_data():
    data = {"balance": balance, "history": history}
    file = open(FILE_NAME, "w")
    json.dump(data, file, indent=4)
    file.close()
def login():
    attempts = 3
    while attempts > 0:
        user = input("Username: ").strip().lower()
        pin = input("PIN: ")
        if user == USERNAME and pin == PIN:
            print("Login successful. Welcome, Muhammad Faizan!")
            return True
        attempts -= 1
        print("Wrong username or PIN. Attempts left:", attempts)
    print("Too many failed attempts. Account locked.")
    return False
def check_balance():
    print("Current balance:", balance)
def deposit():
    global balance
    try:
        amount = float(input("Enter amount to deposit: "))
        if amount <= 0:
            print("Amount must be greater than zero.")
        else:
            balance += amount
            time_now = datetime.now().strftime("%d-%m-%Y %H:%M:%S")
            history.append(time_now + " | Deposit: +" + str(amount))
            print("Deposited:", amount)
            print("New balance:", balance)
            save_data()
    except ValueError:
        print("Invalid input. Please enter a number.")
def withdraw():
    global balance
    try:
        amount = float(input("Enter amount to withdraw: "))
        if amount <= 0:
            print("Amount must be greater than zero.")
        elif amount > balance:
            print("Insufficient balance.")
        else:
            balance -= amount
            time_now = datetime.now().strftime("%d-%m-%Y %H:%M:%S")
            history.append(time_now + " | Withdraw: -" + str(amount))
            print("Withdrawn:", amount)
            print("New balance:", balance)
            save_data()
    except ValueError:
        print("Invalid input. Please enter a number.")
def show_history():
    if len(history) == 0:
        print("No transactions yet.")
    else:
        print("\t Transaction History ")
        for item in history:
            print(item)
def bank_system():
    if not login():
        return
    while True:
        print("\n\t BANK MENU ")
        print("1. Check Balance")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. Transaction History")
        print("5. Exit")
        choice = input("Choose an option (1-5): ")
        if choice == "1":
            check_balance()
        elif choice == "2":
            deposit()
        elif choice == "3":
            withdraw()
        elif choice == "4":
            show_history()
        elif choice == "5":
            print("Thank you for banking with us. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.") 
load_data()
bank_system()
```
